# Synthetic
Compare Ours, Plug-in, DR, Oracle DR, and R.

```python
from pathlib import Path
from ours.demo import select_synthetic_conditions, show_synthetic_results
from ours.synthetic_experiments import run_suite, summarize_results
```

## Settings
`quick`: two small replications. `full`: 100 sample-size replications and 30 sweep replications.

```python
PROFILE = "quick"       # "quick" | "full"
SETTING = "all"         # "all" | "univariate" | "multivariate"
STUDY = "all"           # "all" | "sample_size" | "noise" | "overlap"
                        # "nuisance" | "representation" | "crossfit"
                        # "undersmoothing" | "same_kernel"
N_REPS = None           # None uses the profile default
PLOTS = False
```

## Run

```python
conditions = select_synthetic_conditions(PROFILE, SETTING, STUDY, N_REPS)
print(f"Running {len(conditions)} conditions...")
raw = run_suite(PROFILE, conditions=conditions, progress=False)
assert raw.status.eq("ok").all(), raw.loc[raw.status.ne("ok"), ["condition_id", "method", "error"]]
summary = summarize_results(raw)
RUN_DIR = Path(raw.attrs["run_dir"])
print(f"Saved: {RUN_DIR}")
```

## Results
Mean MISE (SE).

```python
show_synthetic_results(summary, RUN_DIR, plots=PLOTS)
```
