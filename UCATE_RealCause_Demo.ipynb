# RealCause
Compare Ours, Plug-in, DR, and R using Nyström KRR (700 landmarks).

```python
from ours.demo import realcause_run_dir, show_realcause_results
from ours.realcause_experiments import experiment_config, run_samples
```

## Settings
`N_SAMPLES = 30` runs files 0–29. `full` uses all rows; `quick` reduces rows and the ridge grid.

```python
PROFILE = "full"       # "full" | "quick"
N_SAMPLES = 50         # Number of files, not rows
START_SAMPLE = 0       # First file ID (0–99)
PLOTS = False
```

## Run

```python
config = experiment_config(PROFILE, N_SAMPLES, START_SAMPLE)
RUN_DIR = realcause_run_dir(config)
print(f"Running {N_SAMPLES} files...")
results = run_samples(config, checkpoint_path=RUN_DIR / "samples.csv", verbose=False)
print(f"Saved: {RUN_DIR}")
```

## Results
Noise adjustment uses the same files and requires at least three full-profile files.

```python
table = show_realcause_results(results, RUN_DIR, PROFILE, plots=PLOTS)
```
